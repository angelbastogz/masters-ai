### Mutabilidad, alias y copia

Dos nombres pueden apuntar a la misma lista. Para modificar el contenedor por separado, debemos copiarlo.
Si contiene otros objetos mutables, una copia superficial aún puede compartirlos. Este detalle explica muchos errores silenciosos
en pipelines de datos.

```python
original = ["dato-a", "dato-b"]
alias = original
copia = original.copy()

alias.append("dato-c")
copia.append("dato-independiente")

print("original:", original)
print("alias:   ", alias)
print("copia:   ", copia)
print("¿Mismo objeto?", original is alias, original is copia)
```

### Copiar una estructura anidada

`.copy()` hace una **copia superficial**: crea el contenedor externo, pero conserva
las referencias a sus elementos. Si dentro hay una lista, esa lista sigue compartida.

```python
registro_original = {"texto": "hola", "etiquetas": ["saludo"]}
registro_copia = registro_original.copy()
registro_copia["etiquetas"].append("corto")
print(registro_original)
print(registro_original is registro_copia)
print(registro_original["etiquetas"] is registro_copia["etiquetas"])
```

### Bloques, ciclos y decisiones

Los dos puntos abren un bloque y la sangría indica qué instrucciones pertenecen a

Recorre `[None, 0.0, 0.4, 0.95, 0.99]`. Ignora `None` y detente en el primer score mayor o igual a 0.9. Deben mostrarse 0.0, 0.4 y 0.95. ¿Qué se perdería usando `if not score`?

Resuelve el ejercicio en la celda siguiente y conserva las comprobaciones.

```python
valores_busqueda = [None, 0.0, 0.4, 0.95, 0.99]
# Usa for, continue y break.
for score in valores_busqueda:
    if score is None:
        continue
    if score >= 0.9:
        break
    print(score)

print("===")
for score in valores_busqueda:
    if not score:
        continue
    if score >= 0.9:
        break
    print(score)
```

## 4. Ciclos y comprehensions

Un `for` recorre los elementos de un iterable. En una lista puedes recorrer los
valores directamente; no hace falta construir sus índices. Usa `enumerate` cuando
también necesites la posición y `zip` para avanzar por varias secuencias a la vez.
Por defecto, `zip` se detiene al agotarse la secuencia más corta. Si necesitas
comprobar que tengan la misma longitud, usa `zip(..., strict=True)`.

```python
textos = ["muy bueno", "regular", "excelente", "perfecto"]
scores = [0.82, 0.51, 0.96, 1]

for posicion, (texto, score) in enumerate(zip(textos, scores, strict=True), start=1):
    print(f"{posicion}. {texto:<12} → {score:.0%}")
```

Las comprehensions sirven para transformaciones pequeñas y legibles. Si la
lógica requiere varias decisiones o efectos secundarios, un ciclo explícito
suele ser mejor.

```python
scores_altos = [score for score in scores if score >= 0.80]
longitudes = {texto: len(texto) for texto in textos}
iniciales = {texto[0] for texto in textos}

print(scores_altos)
print(longitudes)
print(iniciales)
```

## 5. Funciones: nombrar una transformación

Una función bien enfocada reduce repetición y permite probar una idea de forma
aislada. Sus parámetros son entradas; `return` produce una salida. Una función
sin `return` explícito devuelve `None`.

```python
def normalizar_texto(texto, *, minusculas=True):
    '''Elimina espacios externos y opcionalmente convierte a minúsculas.'''
    resultado = texto.strip()
    if minusculas:
        resultado = resultado.lower()
    return resultado


print(normalizar_texto("  Hola, IA  "))
print(normalizar_texto("  Hola, IA  ", minusculas=False))
```

El `*` vuelve explícitos los argumentos posteriores: debemos escribir
`minusculas=False`. En la semana 2 añadiremos anotaciones de tipo y Pydantic;
hoy nos concentramos en el comportamiento.

### Ejercicio 7 · Promedio sin inventar datos

Escribe `promedio_o_none(valores)`: devuelve el promedio o `None` si la lista está vacía. Comprueba `[0.0]`, `[]` y `[0.5, 1.0]`. ¿Por qué cero y ausencia necesitan resultados distintos?

Resuelve el ejercicio en la celda siguiente y conserva las comprobaciones.

```python
# Define promedio_o_none y escribe tres comprobaciones.
def promedio_o_none(valores):
    if not valores:
        return None
    return sum(valores) / len(valores)

assert promedio_o_none([0.0]) == 0.0
assert promedio_o_none([]) is None
assert promedio_o_none([0.5, 1.0]) == 0.75
```

> Porque cero es un valor, ausencia significa que la lista está vacía y no hay nada que promediar.

### Ordenar usando una función

`sorted` crea una lista nueva; `.sort()` modifica la lista existente. El argumento
`key` recibe una función que calcula el criterio de orden. Pasamos la función
sin llamarla: Python la aplicará a cada elemento.

```python
evaluaciones = [
    {"modelo": "a", "score": 0.7},
    {"modelo": "b", "score": 0.95},
    {"modelo": "c", "score": 0.8},
]

def obtener_score(evaluacion):
    return evaluacion["score"]

ranking = sorted(evaluaciones, key=obtener_score, reverse=True)
print(ranking)
# lambda permite expresar una función pequeña en el mismo lugar.
assert ranking == sorted(evaluaciones, key=lambda e: e["score"], reverse=True)
```

### De contar a agrupar

Un contador responde «¿cuántos?». Una agrupación conserva los elementos asociados
a cada clave. `defaultdict(list)` crea una lista al acceder a una clave ausente.

```python
from collections import defaultdict

conteo_manual = {}
for etiqueta in etiquetas:
    conteo_manual[etiqueta] = conteo_manual.get(etiqueta, 0) + 1
assert conteo_manual == dict(Counter(etiquetas))

por_etiqueta = defaultdict(list)
for registro in [
    {"etiqueta": "positivo", "score": 0.9},
    {"etiqueta": "neutral", "score": 0.6},
    {"etiqueta": "positivo", "score": 0.8},
]:
    por_etiqueta[registro["etiqueta"]].append(registro["score"])

print(dict(por_etiqueta))
```

## 8. POO: objetos que conservan estado

Una **clase** define cómo construir objetos y qué operaciones ofrecen. Cada
**instancia** es un objeto concreto. Sus **atributos** guardan datos y sus
**métodos** son funciones asociadas al objeto.

`__init__` inicializa una instancia; `self` representa esa instancia dentro de los
métodos. Al llamar `contador.agregar()`, Python pasa `contador` como `self`.

```python
class Contador:
    def __init__(self, nombre):
        self.nombre = nombre
        self.total = 0

    def agregar(self):
        self.total += 1

    def reiniciar(self):
        self.total = 0

contador_a = Contador("lote A")
contador_b = Contador("lote B")
contador_a.agregar()
contador_a.agregar()
print(contador_a.nombre, contador_a.total)
print(contador_b.nombre, contador_b.total)
assert contador_b.total == 0
```

```python
class RegistroCompartido:
    registros = []  # Defecto intencional: la lista pertenece a la clase.

    def agregar(self, texto):
        self.registros.append(texto)

compartido_a = RegistroCompartido()
compartido_b = RegistroCompartido()
compartido_a.agregar("dato de A")
print(compartido_b.registros)
assert compartido_a.registros is compartido_b.registros
```

## 9. Caso integrador: un registro de predicciones

La clase siguiente conserva predicciones ya normalizadas y combina estructuras de
datos con comportamiento y estado. Cada registro recibe su posición al agregarse.
`resumen` devuelve cantidad, conteos y promedio; el promedio es `None` cuando no
hay registros.


```python
predicciones = [
    {"texto": "Excelente", "etiqueta": "positivo", "score": 0.98},
    {"texto": "Regular", "etiqueta": "neutral", "score": 0.66},
    {"texto": "Llegó roto", "etiqueta": "negativo", "score": 0.93},
    {"texto": "Cero válido", "etiqueta": "neutral", "score": 0.0},
]
```

```python
class RegistroPredicciones:
    def __init__(self, nombre):
        self.nombre = nombre
        self._registros = []

    def agregar(self, registro, posicion):
        guardado = registro.copy()
        guardado["posicion"] = posicion
        self._registros.append(guardado)

    def average_by_group(self):
        por_etiqueta = defaultdict(list)
        for registro in self._registros:
            por_etiqueta[registro["etiqueta"]].append(registro["score"])
        return {etiqueta: mean(scores) for etiqueta, scores in por_etiqueta.items()}

    def resumen(self):
        scores = [registro["score"] for registro in self._registros]
        return {
            "cantidad": len(self._registros),
            "conteos": dict(Counter(r["etiqueta"] for r in self._registros)),
            "promedio": sum(scores) / len(scores) if scores else None,
        }

    def reiniciar(self):
      self._registros = []


lote = RegistroPredicciones("Demostración")
for posicion, registro in enumerate(predicciones):
    lote.agregar(registro, posicion)
print(lote.resumen())
```

y promedio **0.6425**. Este promedio describe los scores; no mide exactitud del
modelo porque no tenemos etiquetas de referencia.

¿Cómo distinguirías cero de ausencia?
> Cero es un valor válido y la ausencia se representa con `None`. Para distinguirlos uso `valor is None` en lugar de `if not valor`, ya que `if not` trata 0 y 0.0 como falsos y descartaría un score válido.

¿Qué ocurre al copiar un contenedor con elementos mutables?
> `.copy()` hace una copia superficial: crea un contenedor nuevo, pero los elementos internos siguen siendo los mismos objetos. Si modifico una lista o un diccionario dentro de la copia, también cambia el original. Para evitarlo hay que copiar los elementos internos o usar `copy.deepcopy`. Por eso `agregar` hace `registro.copy()` antes de añadir `posicion`, para no modificar el diccionario original de `predicciones`.

¿Cuáles son los atributos y métodos de `RegistroPredicciones`?
> Atributos: `nombre` y `_registros` (el guion bajo indica que es de uso interno).
> Métodos: `__init__`, `agregar`, `average_by_group`, `resumen` y `reiniciar`.

¿Por qué dos lotes deben conservar listas distintas?
> Porque cada lote representa un conjunto de datos independiente. Si compartieran la lista, como en `RegistroCompartido`, una predicción agregada a un lote también aparecería en el otro y sus resúmenes serían incorrectos. Por eso `_registros` se inicializa en `__init__`, así cada instancia tiene su propia lista.
